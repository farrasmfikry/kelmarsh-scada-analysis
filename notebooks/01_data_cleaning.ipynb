```python
import sys
from pathlib import Path
import pandas as pd

# Add root directory to path
sys.path.append(str(Path.cwd().parent))

from src.data_loader import load_scada_data
from src.analytics import generate_scada_overview_report

# 1. Load all available Kelmarsh turbines (T01 through T06)
turbine_dfs = {}
turbine_ids = ['1', '2', '3', '4', '5', '6']

for t_id in turbine_ids:
    file_path = Path(f"../data/Turbine_Data_Kelmarsh_{t_id}_2020-01-01_-_2021-01-01.csv") # Adjust filename pattern as needed
    if file_path.exists():
        df = load_scada_data(file_path)
        turbine_dfs[f"T0{t_id[-1]}"] = df

# 2. Generate the report if data is present
if turbine_dfs:
    report_table = generate_scada_overview_report(turbine_dfs, power_col='Power (kW)', rated_power_mw=2.0)
    
    print("=" * 85)
    print("SCADA DATA OVERVIEW REPORT")
    print("=" * 85)
    print(report_table.to_string())
    print("=" * 85)
else:
    print("No turbine CSV files found in data/ folder. Please check paths.")
```
